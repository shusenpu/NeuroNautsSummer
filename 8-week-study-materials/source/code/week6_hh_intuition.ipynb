# Week 6 HH Intuition

This notebook is not a full HH derivation. It plots a beginner version of sodium and potassium timing.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 20, 1000)
V = -70 + 100*np.exp(-((t-5)/1.0)**2) - 15*np.exp(-((t-8)/1.8)**2)
m = 1/(1+np.exp(-(t-4)))
h = 1/(1+np.exp((t-6)))
n = 1/(1+np.exp(-(t-7)))
g_Na = m**3 * h
g_K = n**4

fig, axes = plt.subplots(3, 1, figsize=(7, 7), sharex=True)
axes[0].plot(t, V)
axes[0].set_ylabel("V (mV)")
axes[1].plot(t, m, label="m: Na activation")
axes[1].plot(t, h, label="h: Na inactivation")
axes[1].plot(t, n, label="n: K activation")
axes[1].legend()
axes[1].set_ylabel("gate value")
axes[2].plot(t, g_Na, label="g_Na")
axes[2].plot(t, g_K, label="g_K")
axes[2].legend()
axes[2].set_ylabel("relative g")
axes[2].set_xlabel("time (ms)")
plt.show()
```
