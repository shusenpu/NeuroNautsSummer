## Setup

Run this cell once. You do not need to edit it.

```python
import numpy as np
import matplotlib.pyplot as plt

DT = 1.0
TIME = np.arange(0, 301, DT)
REST = -65
THRESHOLD = -50
RESET = -68
TAU = 24

plt.rcParams['figure.figsize'] = (9, 4)
plt.rcParams['font.size'] = 12

def run_simple_neuron(input_strength=12, threshold=THRESHOLD, start=50, end=250):
    voltage = np.zeros_like(TIME, dtype=float)
    voltage[0] = REST
    spikes = []
    input_signal = np.zeros_like(TIME, dtype=float)
    input_signal[(TIME >= start) & (TIME <= end)] = input_strength
    pause = 0
    for i in range(1, len(TIME)):
        if pause > 0:
            voltage[i] = RESET
            pause -= 1
            continue
        leak = -(voltage[i-1] - REST) / TAU
        push = input_signal[i] * 0.09
        voltage[i] = voltage[i-1] + leak + push
        if voltage[i] >= threshold:
            spikes.append(TIME[i])
            voltage[i] = RESET
            pause = 3
    return voltage, input_signal, np.array(spikes)

def plot_run(input_strength=12, threshold=THRESHOLD, start=50, end=250, title='Simple neuron run'):
    voltage, input_signal, spikes = run_simple_neuron(input_strength, threshold, start, end)
    fig, ax = plt.subplots()
    ax.plot(TIME, voltage, linewidth=2, label='voltage')
    ax.axhline(threshold, linestyle='--', color='crimson', label='threshold')
    ax.axvspan(start, end, color='gold', alpha=0.18, label='input window')
    if len(spikes):
        ax.scatter(spikes, np.full_like(spikes, threshold), color='black', zorder=3, label='spike times')
    ax.set_title(title)
    ax.set_xlabel('time (ms)')
    ax.set_ylabel('voltage')
    ax.legend(loc='lower right')
    plt.show()
    print('Spike count:', len(spikes))
    return spikes

def summarize_runs(rows):
    print('setting | spike count | first spike | last spike')
    print('-' * 48)
    for label, spikes in rows:
        first = '-' if len(spikes) == 0 else int(spikes[0])
        last = '-' if len(spikes) == 0 else int(spikes[-1])
        print(f'{label:>12} | {len(spikes):>11} | {first:>10} | {last:>9}')
```

## Baseline Check

Predict: What should the baseline graph look like?

Run: Keep the default settings.

Look: Record spike count.

Explain: This baseline is what later changes compare against.

```python
baseline_spikes = plot_run(input_strength=12, threshold=-50, start=50, end=250, title='Baseline run')
```

## Activity 5: Change Input Strength

Predict: If input strength increases, what will happen to spike count?

Run: Compare low, medium, and high input strength.

Look: Which setting changes spike count most clearly?

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
strength_rows = []
for strength in [10, 12, 14]:
    spikes = plot_run(input_strength=strength, threshold=-50, start=50, end=250, title=f'Input strength = {strength}')
    strength_rows.append((f'strength {strength}', spikes))

summarize_runs(strength_rows)
```

## Activity 6: Change Threshold

Predict: If the threshold is higher, what will happen to spike count?

Run: Keep input strength fixed and move only the threshold.

Look: Does the voltage cross the dashed line more or less often?

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
threshold_rows = []
for threshold in [-52, -50, -48]:
    spikes = plot_run(input_strength=12, threshold=threshold, start=50, end=250, title=f'Threshold = {threshold}')
    threshold_rows.append((f'thresh {threshold}', spikes))

summarize_runs(threshold_rows)
```

### Activity 6 Evidence Note

## Activity 7: Change Input Duration

Predict: If input lasts longer, what will happen to spike count?

Run: Keep strength fixed and change only the input window.

Look: Compare the shaded window and spike count.

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
duration_rows = []
for end_time in [150, 200, 250]:
    spikes = plot_run(input_strength=12, threshold=-50, start=50, end=end_time, title=f'Input ends at {end_time} ms')
    duration_rows.append((f'end {end_time}', spikes))

summarize_runs(duration_rows)
```

## Compare The Three Experiments

You now have three controlled comparisons.

Look across your results:

- input strength
- threshold
- input duration

Which one gave the clearest graph evidence?

```python
all_experiments = {
    'input strength': strength_rows,
    'threshold': threshold_rows,
    'input duration': duration_rows,
}

for experiment, rows in all_experiments.items():
    counts = [len(spikes) for label, spikes in rows]
    print(experiment, 'counts:', counts, 'range:', max(counts) - min(counts))
```
