Notebook Four focuses on spike trains.

A spike train is a list of event times.

You will practice:

## Setup

Run this cell once. You do not need to edit it.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (9, 3)
plt.rcParams['font.size'] = 12

def plot_spike_train(spike_times, title='Spike train', window=(0, 200)):
    spike_times = np.array(spike_times)
    plt.figure()
    plt.eventplot(spike_times, lineoffsets=0.5, linelengths=0.6, color='black')
    plt.xlim(window)
    plt.ylim(0, 1)
    plt.yticks([])
    plt.xlabel('time (ms)')
    plt.title(title)
    plt.show()
    print('Spike times:', spike_times.tolist(), 'ms')
    print('Spike count:', len(spike_times))
    return spike_times

def firing_rate(spike_times, window_ms):
    seconds = window_ms / 1000
    return len(spike_times) / seconds

def isi_gaps(spike_times):
    return np.diff(np.array(spike_times))
```

```python
spike_times = plot_spike_train([18, 44, 70, 96, 122, 148], title='Activity 8: count the spikes', window=(0, 160))
```

## Activity 9: Estimate Firing Rate

Predict: Will the firing rate be low, middle, or high?

Run: Use spike count and window length.

Look: Rate = spikes per second.

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
window_ms = 160
rate_hz = firing_rate(spike_times, window_ms)
print('Spike count:', len(spike_times))
print('Window:', window_ms / 1000, 'seconds')
print('Firing rate:', round(rate_hz, 1), 'Hz')
```

## Activity 10: Same Count, Different Timing

Predict: Can two spike trains have the same count but different timing?

Run: Compare the two plots.

Look: Same count, different spacing.

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
regular = plot_spike_train([20, 50, 80, 110, 140], title='Regular timing', window=(0, 160))
irregular = plot_spike_train([12, 25, 82, 91, 145], title='Irregular timing', window=(0, 160))
print('Regular count:', len(regular))
print('Irregular count:', len(irregular))
```

## Activity 11: Interspike Interval

Interspike interval means gap between spikes.

Short name: ISI.

Predict: Are the gaps regular or irregular?

Run: Subtract neighboring spike times.

Look: Compare the gap numbers.

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
example_spikes = np.array([8, 29, 37, 58, 66])
gaps = isi_gaps(example_spikes)
print('Spike times:', example_spikes.tolist(), 'ms')
print('ISI gaps:', gaps.tolist(), 'ms')

plt.figure(figsize=(9, 2.6))
plt.eventplot(example_spikes, lineoffsets=0.5, linelengths=0.6, color='black')
for left, right, gap in zip(example_spikes[:-1], example_spikes[1:], gaps):
    plt.annotate('', xy=(right, 0.85), xytext=(left, 0.85), arrowprops=dict(arrowstyle='<->'))
    plt.text((left + right) / 2, 0.92, f'{gap} ms', ha='center')
plt.yticks([])
plt.xlabel('time (ms)')
plt.title('ISI gaps between neighboring spikes')
plt.show()
```
