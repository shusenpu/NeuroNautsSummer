## Setup

Run this cell once. You do not need to edit it.

```python
import numpy as np
import matplotlib.pyplot as plt

DT = 1.0
TIME = np.arange(0, 301, DT)
REST = -65
THRESHOLD = -50
RESET = -68
TAU = 24

plt.rcParams['figure.figsize'] = (9, 4)
plt.rcParams['font.size'] = 12

def run_simple_neuron(input_strength=12, threshold=THRESHOLD, start=50, end=250):
    voltage = np.zeros_like(TIME, dtype=float)
    voltage[0] = REST
    spikes = []
    input_signal = np.zeros_like(TIME, dtype=float)
    input_signal[(TIME >= start) & (TIME <= end)] = input_strength
    pause = 0
    for i in range(1, len(TIME)):
        if pause > 0:
            voltage[i] = RESET
            pause -= 1
            continue
        leak = -(voltage[i-1] - REST) / TAU
        push = input_signal[i] * 0.09
        voltage[i] = voltage[i-1] + leak + push
        if voltage[i] >= threshold:
            spikes.append(TIME[i])
            voltage[i] = RESET
            pause = 3
    return voltage, input_signal, np.array(spikes)

def plot_run(input_strength=12, threshold=THRESHOLD, start=50, end=250, title='Simple neuron run'):
    voltage, input_signal, spikes = run_simple_neuron(input_strength, threshold, start, end)
    fig, ax = plt.subplots()
    ax.plot(TIME, voltage, linewidth=2, label='voltage')
    ax.axhline(threshold, linestyle='--', color='crimson', label='threshold')
    ax.axvspan(start, end, color='gold', alpha=0.18, label='input window')
    if len(spikes):
        ax.scatter(spikes, np.full_like(spikes, threshold), color='black', zorder=3, label='spike times')
    ax.set_title(title)
    ax.set_xlabel('time (ms)')
    ax.set_ylabel('voltage')
    ax.legend(loc='lower right')
    plt.show()
    print('Spike times:', spikes.tolist())
    print('Spike count:', len(spikes))
    return spikes
```

## First Run: Baseline

Predict: What do you expect the voltage graph to do?

Run: Keep the starting settings.

Look: Count the spikes.

Explain: What did the graph show?

```python
baseline_spikes = plot_run(input_strength=12, threshold=-50, start=50, end=250, title='Notebook One baseline')
```

## Try One Change: Input Strength

Predict: If input gets stronger, will spike count go up, down, or stay the same?

Run: Change only input strength.

Look: Compare spike count.

Explain: When I changed ___, the graph changed by ___. I think this means ___.

```python
weaker_spikes = plot_run(input_strength=10, threshold=-50, start=50, end=250, title='Weaker input')
stronger_spikes = plot_run(input_strength=14, threshold=-50, start=50, end=250, title='Stronger input')

print('Weaker input count:', len(weaker_spikes))
print('Baseline count:', len(baseline_spikes))
print('Stronger input count:', len(stronger_spikes))
```
