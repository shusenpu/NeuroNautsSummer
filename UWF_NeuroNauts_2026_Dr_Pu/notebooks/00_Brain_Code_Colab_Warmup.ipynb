## Step 3: Make a simple plot

A plot turns numbers into a picture. In neuroscience, plots help us see patterns that are hard to see from raw numbers.

```python
# Import NumPy, a library that helps us make and work with number lists.
# We give it the short name np because scientists use it so often.
import numpy as np

# Import Matplotlib, a library that helps us draw plots.
# pyplot is the part of Matplotlib used for everyday plotting.
import matplotlib.pyplot as plt

# Make a list of time points from 0 to 10.
# np.linspace(start, stop, number_of_points) creates evenly spaced numbers.
time = np.linspace(0, 10, 100)

# STUDENT SETTINGS:
# Try changing these two numbers after you run the cell once.
slope = 2
starting_value = 1

# Make a simple signal that changes over time.
# This is not a real brain signal yet; it is a gentle plotting warm-up.
signal = slope * time + starting_value

# Start a new figure and choose its size.
plt.figure(figsize=(8, 4))

# Draw the signal as a line.
# The first input is the x-axis, and the second input is the y-axis.
plt.plot(time, signal, linewidth=3)

# Add labels so a reader knows what the axes mean.
plt.xlabel("Time")
plt.ylabel("Signal value")

# Add a title that says the main idea of the plot.
plt.title("A simple signal can change over time")

# Add a faint grid to make the line easier to read.
plt.grid(alpha=0.3)

# Show the plot below this cell.
plt.show()
```

## Step 5: Mini homework-style exploration

Pick one everyday brain story and make a simple signal for it.

Examples:

- Attention during a 10-minute task.
- Hunger across the morning.
- Confidence while practicing a new skill.
- Stress before and after a test.

Discussion sentence: **My plot is useful because __________, but it leaves out __________.**

```python
# STUDENT SETTINGS:
story_name = "confidence while practicing"
starting_level = 2.0
change_per_minute = 0.6
random_wobble = 0.25

# Keep this seed if you want the same random wobble each time.
rng = np.random.default_rng(3)

# Make a 10-minute timeline.
minutes = np.linspace(0, 10, 100)

# Build a simple story signal with a trend plus a little random wobble.
story_signal = (
    starting_level
    + change_per_minute * minutes
    + random_wobble * rng.normal(size=len(minutes))
)

# Plot the story.
plt.figure(figsize=(8, 4))
plt.plot(minutes, story_signal, linewidth=2.5)
plt.xlabel("Time (minutes)")
plt.ylabel("Model signal")
plt.title(story_name)
plt.grid(alpha=0.3)
plt.show()

# A small numerical readout can help us compare plots.
print("Start:", round(story_signal[0], 2))
print("End:", round(story_signal[-1], 2))
print("Change:", round(story_signal[-1] - story_signal[0], 2))
```
